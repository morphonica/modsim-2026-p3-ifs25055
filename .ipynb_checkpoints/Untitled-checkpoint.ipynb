```python
import simpy
import random
import statistics

# Parameter sistem
JUMLAH_MAHASISWA = 500
JUMLAH_SERVER = 2
WAKTU_LAYAN_MIN = 1
WAKTU_LAYAN_MAX = 3
RATA_KEDATANGAN = 0.5  # rata-rata antar kedatangan (menit)

waktu_tunggu = []
waktu_selesai = []

def mahasiswa(env, nama, server):
    datang = env.now
    with server.request() as req:
        yield req
        tunggu = env.now - datang
        waktu_tunggu.append(tunggu)

        waktu_layanan = random.uniform(WAKTU_LAYAN_MIN, WAKTU_LAYAN_MAX)
        yield env.timeout(waktu_layanan)

        waktu_selesai.append(env.now)

def proses_kedatangan(env, server):
    for i in range(JUMLAH_MAHASISWA):
        env.process(mahasiswa(env, f"Mahasiswa-{i+1}", server))
        waktu_datang = random.expovariate(1 / RATA_KEDATANGAN)
        yield env.timeout(waktu_datang)

# Inisialisasi simulasi
env = simpy.Environment()
server = simpy.Resource(env, capacity=JUMLAH_SERVER)

env.process(proses_kedatangan(env, server))
env.run()

# Hasil simulasi
print("Waktu selesai semua mahasiswa :", max(waktu_selesai))
print("Rata-rata waktu tunggu        :", statistics.mean(waktu_tunggu))
print("Waktu tunggu maksimum         :", max(waktu_tunggu))
```
